## 10.1 ตารางเปรียบเทียบ

| อัลกอริทึม | Best | Average | Worst | Space | Stable | In-place |
|---|---|---|---|---|---|---|
| Bubble sort | \(O(n)\) | \(O(n^2)\) | \(O(n^2)\) | \(O(1)\) | ใช่ | ใช่ |
| Selection sort | \(O(n^2)\) | \(O(n^2)\) | \(O(n^2)\) | \(O(1)\) | ไม่ | ใช่ |
| Insertion sort | \(O(n)\) | \(O(n^2)\) | \(O(n^2)\) | \(O(1)\) | ใช่ | ใช่ |
| Merge sort | \(O(n\log n)\) | \(O(n\log n)\) | \(O(n\log n)\) | \(O(n)\) | ใช่ | ไม่ |
| Quick sort | \(O(n\log n)\) | \(O(n\log n)\) | \(O(n^2)\) | \(O(\log n)\) | ไม่ | ใช่ |
| Heap sort | \(O(n\log n)\) | \(O(n\log n)\) | \(O(n\log n)\) | \(O(1)\) | ไม่ | ใช่ |
| Counting sort | \(O(n+k)\) | \(O(n+k)\) | \(O(n+k)\) | \(O(k)\) | ใช่ | ไม่ |

- **Stable** = ข้อมูลที่มี key เท่ากันคงลำดับเดิม (สำคัญมากเมื่อเรียงหลายคีย์ต่อกัน)
- การเรียงแบบ **เปรียบเทียบ** มีขีดจำกัดล่างที่ \(\Omega(n\log n)\); counting/radix เร็วกว่าได้เพราะไม่เปรียบเทียบ
- `sorted()` ของ Python ใช้ **Timsort** = merge + insertion, stable, เร็วมากกับข้อมูลที่เรียงบางส่วน

```python
# กลุ่ม O(n^2) - เข้าใจง่าย เหมาะกับ n เล็กหรือข้อมูลเกือบเรียงแล้ว
def bubble_sort(a):
    a = a[:]
    n = len(a)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):          # ตัวใหญ่สุด "ลอย" ไปท้ายทุกรอบ
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
                swapped = True
        if not swapped:                     # ถ้าไม่มีการสลับ = เรียงแล้ว -> best case O(n)
            break
    return a

def selection_sort(a):
    a = a[:]
    for i in range(len(a)):
        m = i
        for j in range(i + 1, len(a)):      # หาตัวน้อยสุดในส่วนที่ยังไม่เรียง
            if a[j] < a[m]: m = j
        a[i], a[m] = a[m], a[i]             # จำนวน swap น้อยสุด = O(n)
    return a

def insertion_sort(a):
    a = a[:]
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0 and a[j] > key:        # เลื่อนตัวที่ใหญ่กว่าไปทางขวา
            a[j+1] = a[j]; j -= 1
        a[j+1] = key
    return a

data = [64, 25, 12, 22, 11, 90, 34]
print("bubble   :", bubble_sort(data))
print("selection:", selection_sort(data))
print("insertion:", insertion_sort(data))
```

```python
# กลุ่ม O(n log n) แบบ divide and conquer
def merge_sort(a):
    if len(a) <= 1:
        return a
    mid = len(a) // 2
    left, right = merge_sort(a[:mid]), merge_sort(a[mid:])
    return merge(left, right)

def merge(l, r):
    out, i, j = [], 0, 0
    while i < len(l) and j < len(r):
        if l[i] <= r[j]:                    # <= ทำให้ stable
            out.append(l[i]); i += 1
        else:
            out.append(r[j]); j += 1
    out.extend(l[i:]); out.extend(r[j:])
    return out

def quick_sort(a, lo=0, hi=None):
    # in-place, Lomuto partition, สุ่ม pivot เพื่อเลี่ยง worst case O(n^2)
    import random
    a = a[:] if hi is None else a
    if hi is None:
        hi = len(a) - 1
    def part(lo, hi):
        k = random.randint(lo, hi)
        a[k], a[hi] = a[hi], a[k]
        pivot, i = a[hi], lo
        for j in range(lo, hi):
            if a[j] < pivot:
                a[i], a[j] = a[j], a[i]; i += 1
        a[i], a[hi] = a[hi], a[i]
        return i
    def go(lo, hi):
        while lo < hi:
            p = part(lo, hi)
            if p - lo < hi - p:             # recursion ฝั่งเล็กก่อน -> ลึกไม่เกิน O(log n)
                go(lo, p - 1); lo = p + 1
            else:
                go(p + 1, hi); hi = p - 1
    go(lo, hi)
    return a

print("merge:", merge_sort(data))
print("quick:", quick_sort(data))
```

```python
# กลุ่มที่ไม่ใช้การเปรียบเทียบ (non-comparison sort)
def counting_sort(a):
    # เหมาะกับจำนวนเต็มไม่ติดลบที่ช่วงค่า (k) ไม่กว้าง -> O(n + k)
    if not a: return []
    k = max(a)
    count = [0] * (k + 1)
    for x in a: count[x] += 1
    for i in range(1, k + 1): count[i] += count[i-1]     # prefix sum = ตำแหน่งปลาย
    out = [0] * len(a)
    for x in reversed(a):                                 # ไล่ย้อนเพื่อรักษา stability
        count[x] -= 1
        out[count[x]] = x
    return out

def radix_sort(a):
    # เรียงทีละหลัก โดยใช้ counting sort เป็นตัวช่วย -> O(d * (n + 10))
    if not a: return []
    out = a[:]
    exp = 1
    while max(out) // exp > 0:
        buckets = [[] for _ in range(10)]
        for x in out:
            buckets[(x // exp) % 10].append(x)
        out = [x for b in buckets for x in b]
        exp *= 10
    return out

print("counting:", counting_sort([4, 2, 2, 8, 3, 3, 1]))
print("radix   :", radix_sort([170, 45, 75, 90, 802, 24, 2, 66]))
```

```python
# วัดเวลาเทียบกันจริง (empirical comparison)
import random, time

def bench(fn, a):
    t0 = time.perf_counter(); fn(a); return (time.perf_counter() - t0) * 1000

for n in (500, 2000, 5000):
    a = [random.randint(0, 10_000) for _ in range(n)]
    print(f"n = {n}")
    for name, fn in [("insertion", insertion_sort), ("merge", merge_sort),
                     ("quick", quick_sort), ("sorted()", sorted)]:
        print(f"   {name:>10}: {bench(fn, a):8.2f} ms")
```

```python
# Stability สำคัญเมื่อเรียงหลายคีย์: เรียงคีย์รองก่อน แล้วเรียงคีย์หลัก
students = [("ana", 3.2, "IT"), ("ben", 3.8, "CS"), ("cara", 3.2, "CS"), ("dan", 3.8, "IT")]

by_name = sorted(students, key=lambda s: s[0])              # คีย์รอง: ชื่อ
by_gpa  = sorted(by_name, key=lambda s: -s[1])             # คีย์หลัก: GPA มาก->น้อย
for s in by_gpa:
    print(s)                                                # GPA เท่ากันจะเรียงตามชื่อ
```

### เฉลยบทที่ 10

```python
# ข้อ 1: linked list ไม่มี random access -> quick sort เลือก pivot/partition ยาก
# แต่ merge sort ต่อ pointer ได้ และไม่ต้องใช้อาร์เรย์ช่วย -> O(n log n), space O(log n)
class LNode:
    def __init__(self, v, nxt=None): self.v, self.next = v, nxt

def ll_merge_sort(head):
    if head is None or head.next is None:
        return head
    slow, fast = head, head.next                 # หาจุดกลางด้วย slow/fast
    while fast and fast.next:
        slow, fast = slow.next, fast.next.next
    mid, slow.next = slow.next, None             # ตัดเป็นสองท่อน
    l, r = ll_merge_sort(head), ll_merge_sort(mid)
    dummy = tail = LNode(None)
    while l and r:
        if l.v <= r.v: tail.next, l = l, l.next
        else:          tail.next, r = r, r.next
        tail = tail.next
    tail.next = l or r
    return dummy.next

# ข้อ 2: นับ inversion ระหว่างการ merge
def count_inversions(a):
    def sort_count(a):
        if len(a) <= 1: return a, 0
        m = len(a)//2
        l, cl = sort_count(a[:m]); r, cr = sort_count(a[m:])
        out, i, j, c = [], 0, 0, cl + cr
        while i < len(l) and j < len(r):
            if l[i] <= r[j]: out.append(l[i]); i += 1
            else:
                out.append(r[j]); j += 1
                c += len(l) - i                   # l[i:] ทุกตัวมากกว่า r[j]
        out += l[i:] + r[j:]
        return out, c
    return sort_count(a)[1]

# ข้อ 3: three-way partition -> O(n) time, O(1) space
def sort_012(a):
    lo, mid, hi = 0, 0, len(a) - 1
    while mid <= hi:
        if a[mid] == 0:
            a[lo], a[mid] = a[mid], a[lo]; lo += 1; mid += 1
        elif a[mid] == 1:
            mid += 1
        else:
            a[mid], a[hi] = a[hi], a[mid]; hi -= 1
    return a

# ข้อ 4: binary search หาขอบหน้าต่าง -> O(log n + k)
def k_closest(a, x, k):
    lo, hi = 0, len(a) - k
    while lo < hi:
        m = (lo + hi) // 2
        if x - a[m] > a[m + k] - x: lo = m + 1
        else:                       hi = m
    return a[lo:lo + k]

h = LNode(4, LNode(1, LNode(3, LNode(2))))
h = ll_merge_sort(h)
out = []
while h: out.append(h.v); h = h.next
print("ll_merge_sort:", out)
print("inversions   :", count_inversions([2, 4, 1, 3, 5]))
print("sort_012     :", sort_012([2,0,2,1,1,0]))
print("k_closest    :", k_closest([1,2,3,4,5], 3, 3))
```
